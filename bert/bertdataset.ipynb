```python
import pandas as pd

# Login using e.g. `huggingface-cli login` to access this dataset
splits = {'train': 'data/train-00000-of-00001.parquet', 'validation': 'data/validation-00000-of-00001.parquet', 'test': 'data/test-00000-of-00001.parquet'}
df = pd.read_parquet("hf://datasets/Ajhesh7/nepali_sentiment_dataset/" + splits["train"])
df1 = pd.read_parquet("hf://datasets/Ajhesh7/nepali_sentiment_dataset/" + splits["validation"])
df2 = pd.read_parquet("hf://datasets/Ajhesh7/nepali_sentiment_dataset/" + splits["test"])

df = pd.concat([df, df1, df2], ignore_index=True)
df = df.rename(columns={"Sentiment": "sentiment"})

df = df.rename(columns={"Sentences": "sentences"})
# drop any  1 in sentiment column

```

```python
df = df[df.sentiment != 1]
#change any 2 in sentiment column to 1
df['sentiment'] = df['sentiment'].replace(2, 1)
```

```python
df.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
import pyarrow as pa
import pyarrow.parquet as pq
splits = {'train': 'data/train-00000-of-00001.parquet', 'test': 'data/test-00000-of-00001.parquet'}
df3 = pd.read_parquet("hf://datasets/IRIIS-RESEARCH/Sentiment-Analysis-Nepali/" + splits["train"])
df4 = pd.read_parquet("hf://datasets/IRIIS-RESEARCH/Sentiment-Analysis-Nepali/" + splits["test"])
df = pd.concat([df, df3, df4], ignore_index=True)
```

```python
df.info()
```

```python
#to generate a parquet file from current dataframe
import pyarrow as pa
import pyarrow.parquet as pq

dataset = pa.Table.from_pandas(df)
pq.write_table(dataset, 'nepali_sentiment_dataset.parquet')
```
